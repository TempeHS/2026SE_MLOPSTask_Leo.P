# __Design__

## __Defining the business problem to be solved__

In recent years, the price of buying a home in Sydney has skyrocketed due to a persistent supply-demand imbalance. The strong and ongoing population growth continues to increase demand for residential property. This makes it hard for young people to be able save enough money to purchase and own homes. It is estimated to take up to 20 years for a single person in Sydney to save for a median house deposit without external assistance. The property buyer market is filled not only with people looking to buy a home for the first time, but with foreign and domestic property investors looking to continue building wealth. This adds an extra struggle to young people looking to buy a home, particularly when buying property sold through auctions, making the property market in Sydney extremely competitive since investors are typically able to outbid first home buyers.

### __The business problem:__

*How can young people figure out the expected price of a property in the Greater Sydney area before making the effort to purchase?*

## __Refactoring the business problem into a machine learning problem__

How can a machine learning model use historical property sale data to predict the expected price of a property in the Greater Sydney area?

## __Defining success metrics__

- Accurately predicting the sell price of a property in the Greater Sydney area using inputted features
- Allowing first home buyers to know which property to invest time into buying  


## __Researching available data__

[Kaggle.com](https://www.kaggle.com/) has residential property information web-scraped off [Domain.com.au](https://domain.com.au), joined with suburb data and economic data. The data is saved in the CSV file 2.1.2.domain_properties.csv.

- The data columns are:

| Column | Data |
| ------ | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| Price         | The price the property was sold at.  |
| Date Sold     | The date the property was sold recorded as D/MM/YYYY. |
| Suburb        | The suburb the property is situated in. |
| No. Bath      | Count of how many bathrooms the property has. |
| No. Bed       | Count of how many bedrooms the property has. |
| No. Parking   | Count of how many parking spaces the property has. |
| Property Size | The size of the property measured in squared meters. |
| Type          | The type of building e.g. House, Apartment, Vacant land. |
| Suburb Pop.   | The population of the suburb the property is situated in. |
| SMI           | (Suburb Median Income) The median income of the suburb the property is situated in. |
| Suburb SQKM   | The size of the suburb in square kilometers that the property is situated in. |
| Suburb Lat    | The latitude of the suburb that the property is situated in. |
| Suburb Lng    | The longitude of the suburb that the property is situated in. |
| Suburb Elev.  | The elevation of the suburb that the property is situated in. |
| Cash Rate     | The cash rate at the time the property was sold. |
| PII           | (Property Inflation Index) The residential property price inflation index of the quarter that the property was sold in. |
| KM from CBD   | The distance between the property and the centre of Sydney CBD. |
